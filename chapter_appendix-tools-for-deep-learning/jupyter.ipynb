```python
sales.columns = (
    sales.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
sales = sales.rename(columns= {"date" : "datetime"})
```

```python
sales.head(3)
```

```python
categorical_cols = ['country', 'category', 'sales_manager', 'sales_rep', 'device_type']

sales[categorical_cols] = sales[categorical_cols].astype('category')
```

```python
sales.info()
```

```python
sales['order_value_eur'] = pd.to_numeric(
    sales['order_value_eur'].astype(str).str.replace(',', ''),
    errors='coerce'
)
```

```python
sales['cost'] = pd.to_numeric(
    sales['cost'].astype(str).str.replace(',', ''),
    errors='coerce'
)
```

```python
sales.dtypes
```

```python
sales.to_csv(
    "cleaned_sales_data.csv",
    index=False
)
```
